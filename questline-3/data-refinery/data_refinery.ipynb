```python
import pandas as pd

# Load the original dataset
df = pd.read_csv("StudentsPerformance.csv")

print("Initial Dataset Shape:", df.shape)
```

```python
# Checking for missing values per column
missing_before = df.isnull().sum().sum()
print(f"Total missing values found: {missing_before}")

# If any missing values existed, we would drop or impute them:
# df = df.dropna()
```

```python
# Checking count of duplicate rows
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows found: {duplicate_count}")

# Remove duplicate records
df_cleaned = df.drop_duplicates()

print("Dataset Shape after removing duplicates:", df_cleaned.shape)
```

```python
# Confirming no missing values or duplicates remain
assert df_cleaned.isnull().sum().sum() == 0, "Missing values still exist!"
assert df_cleaned.duplicated().sum() == 0, "Duplicates still exist!"

print("Verification Passed: 0 missing values and 0 duplicates detected.")
```

```python
# Saving the cleaned dataset to CSV
df_cleaned.to_csv("cleaned_students_performance.csv", index=False)
print("Saved clean file as: cleaned_students_performance.csv")
```
