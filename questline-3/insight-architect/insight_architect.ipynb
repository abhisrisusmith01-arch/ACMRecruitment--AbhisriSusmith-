```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Set global visual style
sns.set_theme(style="whitegrid")

# Load the Kaggle Students Performance dataset
# Ensure StudentsPerformance.csv is in the same folder as this notebook
df = pd.read_csv("StudentsPerformance.csv")

# Preview the first 5 rows
df.head()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["math score"], kde=True, color="teal", bins=20)
plt.title("Distribution of Student Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("histogram.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x="test preparation course",
    y="math score",
    palette="Blues_d",
    ci=None,
)
plt.title("Average Math Score by Test Preparation Status")
plt.xlabel("Test Preparation Course")
plt.ylabel("Mean Math Score")
plt.tight_layout()
plt.savefig("bar_chart.png")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
education_order = [
    "some high school",
    "high school",
    "some college",
    "associate's degree",
    "bachelor's degree",
    "master's degree",
]
sns.boxplot(
    data=df,
    x="parental level of education",
    y="reading score",
    order=education_order,
    palette="Set2",
)
plt.title("Reading Scores by Parental Education Level")
plt.xlabel("Parental Level of Education")
plt.ylabel("Reading Score")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig("box_plot.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="reading score",
    y="writing score",
    hue="gender",
    alpha=0.8,
    palette="Set1",
)
plt.title("Reading vs. Writing Scores by Gender")
plt.xlabel("Reading Score")
plt.ylabel("Writing Score")
plt.tight_layout()
plt.savefig("scatter_plot.png")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
score_corr = df[["math score", "reading score", "writing score"]].corr()
sns.heatmap(score_corr, annot=True, cmap="coolwarm", vmin=0, vmax=1, fmt=".2f")
plt.title("Exam Performance Correlation Heatmap")
plt.tight_layout()
plt.savefig("heatmap.png")
plt.show()
```
